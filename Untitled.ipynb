```python
for name, df in [("df1", df1), ("df4", df4), ("df5", df5)]:
    print(f"\n===== {name} =====")

    for col in ["language", "queue", "priority"]:
        print(f"\n{col}:")
        print(df[col].value_counts(dropna=False))
```

```python
# Keep only ticket body and source file
frames = []

for name, df in [("df1", df1), ("df4", df4), ("df5", df5)]:
    temp = df[["body"]].copy()
    temp["source_file"] = name
    frames.append(temp)

all_bodies = pd.concat(frames, ignore_index=True)

# Clean text before comparing: lowercase and collapse extra spaces
all_bodies["body_clean"] = (
    all_bodies["body"]
    .astype(str)
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Count how many files each cleaned body appears in
files_per_body = all_bodies.groupby("body_clean")["source_file"].nunique()

# Bodies found in more than one file
duplicates = files_per_body[files_per_body > 1]

print(f"Number of ticket bodies appearing in more than one file: {len(duplicates)}")

print("\nHow many duplicate bodies appear in each number of files:")
print(files_per_body.value_counts().sort_index())

print("\nExamples of duplicated bodies:")
print(all_bodies[all_bodies["body_clean"].isin(duplicates.index)].head(10))
```

```python
# Keep the first occurrence of each cleaned ticket body
deduplicated = all_bodies.drop_duplicates(subset="body_clean", keep="first")

print(f"Rows before removing duplicates: {len(all_bodies)}")
print(f"Rows after removing duplicates: {len(deduplicated)}")
print(f"Duplicate rows removed: {len(all_bodies) - len(deduplicated)}")
```

```python
print(f"Original rows: {len(all_bodies)}")
print(f"Unique rows after deduplication: {len(deduplicated)}")
print(f"Rows removed: {len(all_bodies) - len(deduplicated)}")

# Confirm no duplicate bodies remain
print(f"Remaining duplicate bodies: {deduplicated['body_clean'].duplicated().sum()}")
```

## Day 1: Dataset notes

### Files with support replies
- `df1`: `aa_dataset-tickets-multi-lang-5-2-50-version.csv`
  - 28,587 tickets
  - Languages: English (`en`) and German (`de`)
  - Has `answer` column with support replies
  - 10 queue labels and 3 priority labels

- `df4`: `dataset-tickets-multi-lang3-4k.csv`
  - 4,000 tickets
  - Languages: English, German, Spanish, French, Portuguese
  - Has `answer` column with support replies
  - 10 queue labels and 3 priority labels

- `df5`: `dataset-tickets-multi-lang-4-20k.csv`
  - 20,000 tickets
  - Languages: English and German
  - Has `answer` column with support replies
  - 10 queue labels and 3 priority labels

### Files without replies
- `df2`: `dataset-tickets-german_normalized.csv`
  - German ticket file
  - No support reply column
  - Ignore for now

- `df3`: `dataset-tickets-german_normalized_50_5_2.csv`
  - German ticket file
  - No support reply column
  - Ignore for now

### Duplicate check
- 8,396 unique ticket bodies appeared in more than one file.
- 8,420 duplicate rows were removed in total.
- Final deduplicated dataset: 44,167 tickets.
- No duplicate cleaned ticket bodies remain.

### Decision
- Use `df1`, `df4`, and `df5` because they contain both ticket text and support replies.
- Ignore `df2` and `df3` for now because they do not contain replies.
- Use the deduplicated dataset of 44,167 tickets for later training and testing.
